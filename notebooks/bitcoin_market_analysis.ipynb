# Bitcoin Market Analysis — Retorno, Volatilidade e Drawdown com Python

**Pergunta central:** como preço, retorno, volatilidade, volume e risco de queda do Bitcoin evoluíram entre março de 2017 e fevereiro de 2022?


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('../data/raw/bitcoin_daily_2017_2022.csv',parse_dates=['Date'])
df.head()
```

## Diagnóstico da base

```python
print(df.shape)
print(df['Date'].min(),df['Date'].max())
print('Ausentes:',df.isna().sum().sum())
print('Datas duplicadas:',df['Date'].duplicated().sum())
```

## Métricas

```python
df=df.sort_values('Date').reset_index(drop=True)
df['daily_return']=df['Adj Close'].pct_change()
df['cumulative_return']=df['Adj Close']/df['Adj Close'].iloc[0]-1
peak=df['Adj Close'].cummax()
df['drawdown']=df['Adj Close']/peak-1
df['rolling_vol_30']=df['daily_return'].rolling(30).std()*np.sqrt(365)
```

## Resultados principais

```python
print(f"Retorno acumulado: {df['cumulative_return'].iloc[-1]:.2%}")
print(f"Volatilidade anualizada: {df['daily_return'].std()*np.sqrt(365):.2%}")
print(f"Drawdown máximo: {df['drawdown'].min():.2%}")
```

## Síntese

- Retorno acumulado superior a 4.500%.
- Volatilidade anualizada próxima de 80%.
- Drawdown máximo superior a 80%.
- 2017 e 2022 são anos parciais na base.
- A análise é histórica e não constitui recomendação de investimento.
